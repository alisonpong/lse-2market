### Import & Data Clean for ad_data

```python
# Import the ad_data csv file and sense check the data
ad_data = pd.read_csv('./data/ad_data.csv')
ad_data
```

```python
# Function for data sense check 
def sense_check(df):
    print('### Data Structure ###')
    df.info()
    
    print('\n' + '-'*50 + '\n')

    print('### Statistical Summary ###')
    print(df.describe())

    print('\n' + '-'*50 + '\n')

    print('### Duplicates ###')
    print(df.duplicated().sum())
```

```python
sense_check(ad_data)
```

```python
# Revise column names to lowercases
ad_clean = (
    ad_data
        .rename(columns={
    'ID': 'id',
    'Bulkmail_ad':'bulkmail',
    'Twitter_ad':'twitter',
    'Instagram_ad':'instagram',
    'Facebook_ad':'facebook',
    'Brochure_ad':'brochure'
        })
        .sort_values(by='id')
        .reset_index(drop=True)
)
# Confirm changes
ad_clean.head()
```

### Import & Data Clean for ad_data

```python
# Import the marketing_data csv file and sense check the data
md = pd.read_csv('./data/marketing_data.csv')
md
```

```python
sense_check(md)
```

```python
# Revise marketing_data column names to lowercases
md_1 = (
    md
        .rename(columns={
            'ID':'id',
            'Kidhome':'kid_home',
            'Teenhome':'teen_home',
            'Dt_Customer':'registration_date',
            'AmtLiq':'amt_liquor',
            'AmtVege':'amt_vegetable',
            'AmtNonVeg':'amt_meat',
            'AmtPes':'amt_fish',
            'AmtChocolates':'amt_chocolate',
            'AmtComm':'amt_commodity',
            'NumDeals':'num_deals',
            'NumWebBuy':'num_web_purchase',
            'NumWalkinPur':'num_instore_purchase',
            'NumVisits':'web_visits_per_month',
            'Count_success':'lead_conversion'
        })
        .rename(columns=str.lower)
        .sort_values(by='id')
        .reset_index(drop=True)
)
# Confirm changes
md_1.head()
```

```python
# Writing a function to query for each series' unique value
def check_unique_values(df):
    for col in df.columns:
        print(f"Unique values in {col}:")
        print(df[col].unique())
        print("-" * 20)
```

```python
check_unique_values(md_1)
```

```python
# Removing symbols from the income column values and converting it to an int data type
md_1['income'] = (
    md_1['income']
    .str.replace(r'[^\w\s]', '', regex=True)
    .astype(int)
)
print(md_1['income'].dtype)
md_1.head()
```

```python
# Standardize the education column values
# Define the new category labels
education_categories = {
    'Basic':'High School / Below',
    'Graduation':'Bachelor',
    'Master':'Master',
    '2n Cycle':'Master',
    'PhD':'Doctorate'
}

# Map old values to new values
md_1['education'] = md_1['education'].map(education_categories)

# View changes
md_1['education'].unique()
```

```python
# Standardize the marital status column values
# Define the new marital status category labels
marital_categories = {
    'Single':'Single',
    'Widow':'Single',
    'Divorced':'Single',
    'Alone':'Single',
    'Married':'Couple',
    'Together':'Couple',
    'YOLO':'Other',
    'Absurd':'Other'
}

# Map old values to new values
md_1['marital_status'] = md_1['marital_status'].map(marital_categories)

# View changes
md_1['marital_status'].unique()
```

```python
# Review the number of Other values
# If the number is small, we can consider removing it from the dataset
print((md_1['marital_status'] == 'Other').sum())
```

```python
# Create a new dataframe without the Other value in marital status
md_2 = md_1[md_1['marital_status'] != 'Other'].copy()

# View changes
print(md_2['marital_status'].unique())
print('\n' + '-'*50 + '\n')
md_2
```
